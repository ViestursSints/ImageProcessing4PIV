## Run PIV for a number of micro-PIV slices (refer to README for naming)

```python
import tqdm
import numpy as np
import multiprocessing

def piv_wrapper(args):
    return piv_f.piv_row_parallel_skip(*args)

core_n = 10

im_folder = "/image/folder"
windows = (128, 64)
overlaps = (64, 32)

iter_n = len(windows)
std_thresh = 3
med_thresh = 1

from_slice = 1; slice_to = 23; slice_n = slice_to - from_slice + 1

slice_pile = np.linspace(from_slice,slice_to,num=slice_n)

settings_X = [(im_folder, int(slith), windows, overlaps,
                                  iter_n, std_thresh, med_thresh) for slith 
                                  in range(from_slice, slice_to + 1)]

pool = multiprocessing.Pool(core_n)
#pool.starmap(piv_f.piv_row_parallel_skip, settings_X)

for _ in tqdm.tqdm(
    pool.imap_unordered(piv_wrapper, settings_X),
    total=len(settings_X),
    desc="Processing slices"
):
    pass

pool.close()
pool.join()
```
